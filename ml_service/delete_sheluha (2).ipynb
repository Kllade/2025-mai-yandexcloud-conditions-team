```python
json_dir = 'C:\\Users\\Fedor\\Downloads\\Telegram Desktop\\'
json_files = list(Path(json_dir).glob('data2024_4.json'))
print(json_files)
all_data = []
for json_file in json_files:
    with open(json_file, 'r', encoding='utf-8') as file:
        data = json.load(file)
        if isinstance(data, list):
            all_data.extend(data)
        else:
            all_data.append(data)
df = pd.DataFrame(all_data)
```

```python
df1 = []
```

```python
for text in tqdm(df['text']):
    op = model.run(f"""напиши, является ли это сообщение вопросом который связан с вузом или поступлением в вуз
    или связано с учебным процессом в маи: {text}. Если это не вопрос, то напиши следующее: 
    Это сообщение не является вопросом. 
    Если это сообщение связано с вопросом, то выведи его суть, содержание, остальное убери""")
    if op.text != 'Это сообщение не является вопросом.':
        df1.append(op.text)
```

```python
with open("file.txt", "w") as output:
    for i in df1:
        output.write(str(i))
```
